**Measures of Spread — Variance and Standard Deviation**

We already discovered an important idea:

Two datasets can have the same mean but very different amounts of spread.

Let's understand how statistics measures that spread.

1. Start with our two datasets

Let's use simple datasets:

A = [45, 48, 50, 52, 55]
B = [10, 30, 50, 70, 90]

Both have a mean of 50.

But look at them:

A:
45 → 48 → 50 → 52 → 55

B:
10 → 30 → 50 → 70 → 90

Dataset A is tightly packed around 50.

Dataset B is much more spread out.

**Range**

The simplest measure of spread is the range.

Formula
$$ Range = Maximum - Minimum $$

For A:

$$ 55 - 45 = 10 $$

For B:

90−10=80

| Dataset | Mean | Range |
| ------- | ---: | ----: |
| A       |   50 |    10 |
| B       |   50 |    80 |

The range tells us that B is much more spread out.
But range has a weakness: **it only looks at the minimum and maximum.**

For example:
A = [49, 50, 51, 52, 53]
B = [10, 50, 50, 50, 100]

Both have a range of 90? Actually:

A → 53 − 49 = 4
B → 100 − 10 = 90

More importantly, range doesn't tell us **what happens to all the observations between the minimum and maximum**.

That's where variance comes in


**Variance**

Variance asks: **How far are the individual values from the mean?**

Let's use a tiny dataset: [40, 50, 60]

Mean: 40+50+60 / 3 = 50

Now calculate each value's distance from the mean.
| Value | Mean | Difference |
| ----: | ---: | ---------: |
|    40 |   50 |        -10 |
|    50 |   50 |          0 |
|    60 |   50 |        +10 |


We have:

-10, 0, +10

If we simply average these differences:

$$ \frac{-10+0+10}{3}=0 $$

That's a problem!

The data clearly has some spread, but the positive and negative differences **cancel each other out**.



**Why do we square the differences?**

Statistics solves this by squaring each difference.

(-10)² = 100
(0)²   = 0
(+10)² = 100

Now:

$$ 100,\ 0,\ 100 $$

Average them:

$$ \frac{100+0+100}{3}=66.67 $$

This is the variance for this population.

So conceptually:

**Variance = average of the squared distances from the mean.**


**But variance has a problem**

Suppose our data is people's age.

Age is measured in:

years

But variance is measured in:

years²

For example:

Variance = 800 years²

That's not very intuitive.

We would rather have a measure of spread in the **same units as the original data.**

That's why we have:

**Standard Deviation**

Standard deviation is simply:

$$ SD = \sqrt{Variance} $$


          

So if:

$$ Variance = 800 $$

then:

$$ SD = \sqrt{800} $$

approximately:

$$ 28.28 $$

Now the result is back in the original unit.

If the data is measured in years, standard deviation is also measured in years.

```python
import numpy as np

A = np.array([48, 49, 50, 51, 52])
B = np.array([10, 30, 50, 70, 90])

print("Range of A",np.ptp(A))
print("Range of B",np.ptp(B))

print("A variance:", np.var(A))
print("B variance:", np.var(B))


print("A SD:", np.std(A))
print("B SD:", np.std(B))
```

**Lesson 6: Percentiles, Quartiles, and Interquartile Range (IQR)**

What is a percentile?

**A percentile tells us the relative position of a value in a dataset.**

For example, the 80th percentile is a value at or below which roughly 80% of observations fall, depending on the percentile calculation method.

Imagine a student scores at the 80th percentile on a test. This means their result is higher than or equal to approximately 80% of the comparison group, under the usual percentile-rank interpretation.

It does not mean the student scored 80 marks out of 100.


What are quartiles?

Quartiles divide an ordered dataset into four parts. Each part represents approximately 25% of the observations.

There are three main quartiles:

Q1

**First quartile — 25th percentile**

About 25% of observations lie at or below this value.

Q2

**Second quartile — 50th percentile**

This is the median, the middle of the ordered dataset.

Q3

**Third quartile — 75th percentile**

About 75% of observations lie at or below this value.

Think of quartiles as checkpoints along your ordered data:
* Q1 marks the 25% point.
* Q2 marks the 50% point.
* Q3 marks the 75% point.


**What is the Interquartile Range (IQR)?**

Now we get to a very useful concept.

The interquartile range (IQR) measures the spread of the middle 50% of your data.

Its formula is:

IQR=Q3−Q1
	​


In simple terms:

Q1 marks the lower boundary of the middle half.

Q3 marks the upper boundary of the middle half.

Subtract them to measure the width of that middle half.


**Why do we need IQR if we already have standard deviation?**

Remember what you learned about outliers? The mean and standard deviation can be strongly affected by extreme values.

The IQR focuses on the middle 50% of the observations, so it is generally more resistant to extreme values.

                   
                    Measure	               What it tells us
                    Range	               Distance between the minimum and maximum
                    Variance	           Average squared deviation from the mean
                    Standard deviation	   Spread measured in the original units
                    IQR	                   Spread of the middle 50%


**Use IQR to detect potential outliers**

Let's revisit a dataset from our previous lesson:

data = [20, 21, 22, 23, 24, 100]

Most values are between 20 and 24, but 100 is much larger.

We can use the IQR rule to flag potential outliers.

The rule uses two boundaries, called fences:

**Lower fence=Q1−1.5(IQR)**

**Upper fence=Q3+1.5(IQR)**

**Any observation below the lower fence or above the upper fence is flagged as a potential outlier.**


```python
import numpy as np

marks = [20, 30, 35, 40, 45, 50, 55, 60, 70, 95]

q1 = np.percentile(marks, 25)
q2 = np.percentile(marks, 50)
q3 = np.percentile(marks, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

```python
import numpy as np

data = np.array([20, 21, 22, 23, 24, 100])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower fence:", lower_fence)
print("Upper fence:", upper_fence)
```

```python
import numpy as np

data = np.array([5, 10, 15, 20, 25, 30, 35, 40])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("q1",q1)
print("q2",q2)
print("q3",q3)
print("iqr",iqr)
```

```python
import numpy as np

data = np.array([12, 14, 15, 16, 18, 19, 20, 85])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

upper_fence = q3 + (1.5 * iqr)

print("q1",q1)
print("q3",q3)
print("iqr",iqr)
print("upper fence",upper_fence)
```

**Lesson 7: Correlation — Understanding Relationships Between Variables**

Let's learn correlation from the basics, using simple examples and Python. You don't need to learn probability or advanced mathematics first.

1. **What is correlation?**

**Correlation measures how two variables change in relation to each other.**


The three main types of correlation

**Positive correlation:** When one variable increases, the other tends to increase too.

    Example: More hours studied, generally higher marks.

**Negative correlation:** When one variable increases, the other tends to decrease.

    Example: As product prices rise, the number of units sold may fall.

**Little or no correlation:**  There is no clear linear relationship between the variables.

    Example: Shoe size and exam marks would generally not be expected to have a meaningful linear relationship.


**The correlation coefficient**: A commonly used measure is the Pearson correlation coefficient, represented by **r**.
Its value always lies between −1 and +1.

-1	     Perfect negative linear correlation (As one variable increases,
            the other decreases in a perfectly straight-line pattern.)
0	     No linear correlation (No straight-line relationship.
            A nonlinear relationship could still exist.)
+1	     Perfect positive linear correlation (Both variables follow a perfectly
            straight-line pattern in the same direction.)



Absolute value of r   Common interpretation

0.00–0.19        Very weak

0.20–0.39    	 Weak

0.40–0.59		 Moderate

0.60–0.79		 Strong

0.80–1.00		 Very strong

These labels are only rules of thumb; the context matters. A correlation of r=0.75 is strongly positive, while r=−0.75 is strongly negative.

```python
import numpy as np

hours_studied = [1, 2, 3, 4, 5]
exam_marks = [40, 50, 60, 70, 80]

correlation = np.corrcoef(
    hours_studied,
    exam_marks
)[0, 1]

print(correlation)
```

**What does [0, 1] mean?**

np.corrcoef() returns a correlation matrix:

[[1.0,  correlation],
 [correlation, 1.0]]

[0, 0] is the correlation of the first variable with itself: 1.

[1, 1] is the correlation of the second variable with itself: 1.

[0, 1] is the correlation between the two different variables.



**The key concepts to remember**

* **Positive correlation**: variables tend to move in the same direction.

* **Negative correlation**: variables tend to move in opposite directions.

* **Correlation near zero**: little or no linear relationship; a nonlinear relationship may still exist.

* **Correlation does not imply causation**: a relationship alone doesn't establish that one variable causes the other.

**When correlation can be misleading**

Let's explore a particularly interesting example: two variables can have a strong relationship that Pearson correlation fails to capture.

**Pearson correlation** measures linear association, so it can miss relationships that follow a curve.

Next, we'll look at:

Why a curved relationship can have a Pearson correlation near zero.

How a single outlier can change correlation.

How to calculate and interpret correlation on a realistic Python dataset.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([-3, -2, -1, 0, 1, 2, 3])
y = x**2

r = np.corrcoef(x, y)[0, 1]

print("Correlation:", r)

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y = x²")
plt.title("A curved relationship")
plt.show()
```

**An outlier can change the correlation dramatically**

Suppose we record study hours and exam marks.

Initially, the data follows a perfect straight-line relationship.

```python
import numpy as np

hours = np.array([1, 2, 3, 4, 5])
marks = np.array([2, 4, 6, 8, 10])

print(np.corrcoef(hours, marks)[0, 1])
# 1.0
```

Now imagine adding one unusual observation: someone studies for 6 hours but receives a score of 0.

```python
import numpy as np
import matplotlib.pyplot as plt

hours = np.array([1, 2, 3, 4, 5, 6])
marks = np.array([2, 4, 6, 8, 10, 0])

print(np.corrcoef(hours, marks)[0, 1])
# Approximately 0.143
```

The correlation drops from **1.0 to approximately 0.143.**

This is because Pearson's correlation is sensitive to unusual observations, especially those that strongly influence the pattern.

An outlier isn't necessarily an error. It could be a genuine observation, so investigate it rather than automatically deleting it

**Lesson 8: Sampling and Sampling Bias**
1. What is a population?

A population is the entire group you want to understand.

For example, imagine you want to understand the monthly salaries of all employees in a company with 10,000 employees.

Population: all 10,000 employees.

Sample: 200 employees selected from those 10,000.

Variable: monthly salary.

Goal: use the sample to understand salaries across the company.

2. **What makes a good sample?**

A good sample represents the population reasonably well.

Suppose a company has:

5,000 junior employees

3,000 mid-level employees

2,000 senior employees

If you select 200 employees entirely from the senior group, your sample may not represent the company as a whole.

A more representative sample would include employees from all three groups in appropriate proportions.

3. **What is sampling bias?**

Sampling bias occurs when the method used to select a sample systematically favors some members of the population over others

Example: You want to know whether employees are satisfied with their company's workplace.

You survey only employees who attend an optional company event. People who attend might be more engaged than those who don't, so the results could give a misleading picture of overall satisfaction.

**The key lesson is that a large sample is not necessarily a representative sample.**

```python
import numpy as np

salaries = np.array([
    20, 22, 24, 25, 27,
    30, 32, 35, 60, 80
])

print("Population mean:", np.mean(salaries))
print("Population median:", np.median(salaries))

sample_a = salaries[:5]
sample_b = salaries[[0, 2, 4, 6, 9]]

print("Sample A mean:", np.mean(sample_a))
print("Sample B mean:", np.mean(sample_b))
```

**Lesson 9: The Normal Distribution.**
1. What is a normal distribution?
A normal distribution is a pattern in which data values cluster around the mean, and values become less common as you move farther away from the mean in either direction.

For a normal distribution, three important facts are:
     
        
        **Measure	            What it tells you**
        Mean	                The center of the distribution
        Median	                The middle value; equal to the mean in an ideal normal distribution
        Standard deviation	    How spread out the values are

A smaller standard deviation produces a narrower bell curve. A larger standard deviation produces a wider, flatter curve.


**The 68–95–99.7 rule**
For an ideal normal distribution:
- About 68% of values fall within 1 standard deviation of the mean.
- About 95% fall within 2 standard deviations.
- About 99.7% fall within 3 standard deviations.

Suppose exam marks follow a normal distribution with a mean of 70 and a standard deviation of 10.

            **Exam score example**
    Mean              Standard deviation
    70                10

Approximately 68% of scores fall between 60 and 80.
Approximately 95% fall between 50 and 90.
Approximately 99.7% fall between 40 and 100.

These percentages apply when the distribution is approximately normal, not to every dataset.

**Understanding Z-Scores.**
A z-score tells you how far a value is from the mean, measured in standard deviations.

**1. Understand the idea**
Suppose an exam has:
- Mean score = 70
- Standard deviation = 10
- Your score = 85
Your score is 15 marks above the mean. Since one standard deviation is 10 marks, your score is **1.5 standard deviations above the mean.**
The formula is:
\[
z=\frac{x-\mu}{\sigma}



**2. Interpret positive, negative, and zero z-scores**
                	       
        Z-Score	      Meaning
        +2	        Two standard deviations above the mean
        +1	        One standard deviation above the mean
        0	        Exactly at the mean
        -1	        One standard deviation below the mean
        -2	        Two standard deviations below the mean
        

A z-score describes a value's position relative to its distribution. A negative z-score does not mean the original value is negative.

```python
#Calculate z-scores
import numpy as np

mean = 70
std_dev = 10
score = 85

z_score = (score - mean) / std_dev

scores = np.array([50, 60, 70, 80, 90])

z_scores = (scores - 70) / 10

print(z_scores)

print("Z-score:", z_score)
# 1.5
```

**4. Why are z-scores useful?**
Imagine two students take different exams:
- Student A scores 85 on an exam with mean 70 and standard deviation 10.
- Student B scores 90 on an exam with mean 80 and standard deviation 5.
Student A's z-score is \(1.5\).
Student B's z-score is \(2\).
Although Student B scored only 5 marks higher than Student A, Student B performed farther above their exam's mean relative to the variation in scores.
Z-scores help us compare relative positions across different scales.


**Using Z-Scores to Identify Unusual Values.**

A z-score tells us how far a value is from the mean in standard deviation units.
For data that follows an approximately normal distribution, a common rule of thumb is:
                        
                    
    Absolute z-score		Interpretation
    Less than 2	            Not particularly unusual
    2 to less than 3	    Unusual; worth investigating
    3 or more	            Very unusual; investigate further

These are guidelines, not strict rules. A large z-score doesn't automatically mean the observation is wrong.

```python
import numpy as np

scores = np.array([65, 70, 72, 68, 75, 71, 69, 110])

mean = np.mean(scores)
std_dev = np.std(scores)

z_scores = (scores - mean) / std_dev

print("Mean:", mean)
print("Standard deviation:", std_dev)
print("Z-scores:", np.round(z_scores, 2))

unusual = scores[np.abs(z_scores) >= 2]
print("Values with |z| >= 2:", unusual)
```

**Z-score Outlier Detection**
Let’s learn how to use z-scores to detect unusual values in a dataset, then compare this method with the IQR method you learned earlier.

1. What is an outlier?

An outlier is a value **that is unusually far from the other values in a dataset**.

Imagine these exam scores:
48, 50, 51, 49, 52, 50, 150
Most scores are between 48 and 52, but one score is 150. That value looks unusual.

A z-score helps us measure how far a value is from the mean, in standard deviations.



    			$$	\[
				z=\frac{x-\mu}{\sigma}
				\] $$

- \(x\): the value we are checking
- \(\mu\): the dataset's mean
- \(\sigma\): the dataset's standard deviation


A large positive or negative z-score means the value is far from the mean.


```python
# 2. Calculate z-scores using NumPy


import numpy as np

data = np.array([48, 50, 51, 49, 52, 50, 150])

mean = np.mean(data)
std_dev = np.std(data)

z_scores = (data - mean) / std_dev

print("Mean:", mean)
print("Standard deviation:", std_dev)
print("Z-scores:", np.round(z_scores, 2))


outliers = data[np.abs(z_scores) > 2]

print("Potential outliers:", outliers)
```

What is happening in the above program ?
1. np.mean(data) calculates the average.
2. np.std(data) calculates the population standard deviation by default.
3. data - mean subtracts the mean from every value.
4. Dividing by std_dev converts each difference into a z-score.

The mean is approximately 64.29, and the standard deviation is approximately 35.01. Notice that the unusual value, 150, also pulls the mean and standard deviation upward.


**3. Identify potential outliers**

A common rule of thumb is:

Absolute z-score		Interpretation
* Less than 2		    Not particularly unusual by this rule
* 2 to less than 3		Potentially unusual
* 3 or greater		    Very unusual


These are guidelines, not strict laws. Let's flag values whose absolute z-score is greater than 2.

```python
# 4. Perform the same analysis using pandas
# Pandas is particularly useful when working with real datasets stored in tables.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "score": [48, 50, 51, 49, 52, 50, 150]
})

df["z_score"] = (
    df["score"] - df["score"].mean()
) / df["score"].std(ddof=0)

df["potential_outlier"] = df["z_score"].abs() > 2

print(df)




plt.boxplot(df)
plt.ylabel("Score")
plt.title("Box Plot of Scores")
plt.show()
```

The table above will contain three columns:
- score: the original observation
- z_score: how far the score is from the mean, in standard deviations
- potential_outlier: True when the absolute z-score exceeds 2


Here, ddof=0 ensures we use the population standard deviation, matching NumPy's default.

**Central Limit Theorem (CLT) :**

The Central Limit Theorem is important because it helps us understand how sample averages behave, even when the original data is not normally distributed.

Understand the basic idea:

Imagine a company wants to understand the salaries of all its employees.

Step 1: Original population
Employee salaries might be strongly right-skewed because a small number of employees earn much higher salaries than others.


Step 2: Take many random samples
Take a sample of 30 employees, calculate their average salary, and repeat this process many times.


Step 3: Look at all the averages
Plot the averages from all those samples. Under suitable conditions, their distribution tends to look approximately normal.

**2. What does the theorem actually say?**

Under suitable conditions, when we take sufficiently large random samples from a population and calculate each sample's mean, the distribution of those sample means becomes approximately normal.

There are three important things to remember:
- The original population does not have to be normally distributed.
- We calculate the mean of each sample, not the mean of every individual observation in the population.
- The distribution of those sample means tends to become more normal as the sample size increases, provided the conditions for the theorem are met.
  
The required sample size depends on the shape of the original distribution and other factors. A sample size of 30 is a common guideline, not a universal guarantee.

```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Create a right-skewed population
population = np.random.exponential(
    scale=50,
    size=100_000
)

# Draw 5,000 samples, each containing 30 values
sample_means = []

for _ in range(5000):
    sample = np.random.choice(
        population,
        size=30,
        replace=True
    )

    sample_means.append(np.mean(sample))

# Plot the original population
plt.figure()
plt.hist(population, bins=60, edgecolor="black")
plt.title("Original Population: Right-Skewed")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

# Plot the sample means
plt.figure()
plt.hist(sample_means, bins=50, edgecolor="black")
plt.title("Distribution of 5,000 Sample Means")
plt.xlabel("Sample mean")
plt.ylabel("Frequency")
plt.show()
```

Run the above code and compare the two histograms.

You should see that the original population is right-skewed, while the distribution of sample means is much more bell-shaped.

**4. Why is this useful?**
Suppose you cannot measure every employee's salary, every customer's spending, or every product's lifetime.
You can take samples and calculate their averages. The CLT helps explain why those averages have a predictable distribution, which is a foundation for estimating population averages and constructing confidence intervals.

One important distinction:

**The CLT does not say that the original data becomes normal. It says that the distribution of sample means tends to become approximately normal.**

The key takeaway is that **the Central Limit Theorem applies to the distribution of sample means, not necessarily the original data**.
